### 4. Perbandingan Performa Team

```python
import pandas as pd
import plotly.express as px

# Konversi MatchDate ke datetime jika belum
EPL['MatchDate'] = pd.to_datetime(EPL['MatchDate'])

# Ambil semua musim unik
seasons = EPL['Season'].unique()
all_stats = []

# Loop untuk setiap musim
for season in seasons:
    df_season = EPL[EPL['Season'] == season]
    teams = pd.unique(df_season[['HomeTeam', 'AwayTeam']].values.ravel())

    for team in teams:
        team_df = df_season[(df_season['HomeTeam'] == team) | (df_season['AwayTeam'] == team)].copy()

        # Hitung metrik performa
        team_df['Goals'] = team_df.apply(lambda row: row['FullTimeHomeGoals'] if row['HomeTeam'] == team else row['FullTimeAwayGoals'], axis=1)
        team_df['Fouls'] = team_df.apply(lambda row: row['HomeFouls'] if row['HomeTeam'] == team else row['AwayFouls'], axis=1)
        team_df['Yellow'] = team_df.apply(lambda row: row['HomeYellowCards'] if row['HomeTeam'] == team else row['AwayYellowCards'], axis=1)
        team_df['Red'] = team_df.apply(lambda row: row['HomeRedCards'] if row['HomeTeam'] == team else row['AwayRedCards'], axis=1)

        stats = {
            'Season': season,
            'Team': team,
            'Avg Goals': team_df['Goals'].mean(),
            'Avg Fouls': team_df['Fouls'].mean(),
            'Avg Yellow': team_df['Yellow'].mean(),
            'Avg Red': team_df['Red'].mean()
        }
        all_stats.append(stats)

# Konversi ke DataFrame
radar_df = pd.DataFrame(all_stats)
```

### MODEL TRAINING DBSCAN

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# --- Langkah 1: Salin dan encode data asli ---
EPL_cleaned = EPL.copy()

# Ubah semua kolom bertipe object menjadi bentuk numerik
label_encoders = {}
for col in EPL_cleaned.select_dtypes(include='object').columns:
    # Check if the column exists in the dataframe before encoding
    if col in EPL_cleaned.columns:
        le = LabelEncoder()
        # Handle potential missing values before encoding
        EPL_cleaned[col] = EPL_cleaned[col].astype(str) # Convert to string to handle mixed types
        EPL_cleaned[col] = le.fit_transform(EPL_cleaned[col])
        label_encoders[col] = le  # Simpan encoder kalau mau inverse transform nanti

# --- Langkah 2: Agregasi performa HOME ---
home_stats = EPL_cleaned.groupby(['Season', 'HomeTeam']).agg({
    'FullTimeHomeGoals': 'mean',
    'HomeShots': 'mean',
    'HomeShotsOnTarget': 'mean',
    'HomeCorners': 'mean',
    'HomeFouls': 'mean',
    'HomeYellowCards': 'mean',
    'HomeRedCards': 'mean'
}).reset_index().rename(columns={'HomeTeam': 'Team'})

# --- Langkah 3: Agregasi performa AWAY ---
away_stats = EPL_cleaned.groupby(['Season', 'AwayTeam']).agg({
    'FullTimeAwayGoals': 'mean',
    'AwayShots': 'mean',
    'AwayShotsOnTarget': 'mean',
    'AwayCorners': 'mean',
    'AwayFouls': 'mean',
    'AwayYellowCards': 'mean',
    'AwayRedCards': 'mean'
}).reset_index().rename(columns={'AwayTeam': 'Team'})

# --- Langkah 4: Gabungkan home & away performance ---
team_stats = pd.merge(home_stats, away_stats, on=['Season', 'Team'], how='outer').fillna(0) # Use outer join and fill NaNs

# Hitung rata-rata gabungan
team_stats['AvgGoals'] = (team_stats['FullTimeHomeGoals'] + team_stats['FullTimeAwayGoals']) / 2
team_stats['AvgShots'] = (team_stats['HomeShots'] + team_stats['AwayShots']) / 2
team_stats['AvgShotsOnTarget'] = (team_stats['HomeShotsOnTarget'] + team_stats['AwayShotsOnTarget']) / 2
team_stats['AvgCorners'] = (team_stats['HomeCorners'] + team_stats['AwayCorners']) / 2
team_stats['AvgFouls'] = (team_stats['HomeFouls'] + team_stats['AwayFouls']) / 2
team_stats['AvgYellowCards'] = (team_stats['HomeYellowCards'] + team_stats['AwayYellowCards']) / 2
team_stats['AvgRedCards'] = (team_stats['HomeRedCards'] + team_stats['AwayRedCards']) / 2

# --- Langkah 5: Ambil fitur numerik untuk pemodelan ---
features = ['AvgGoals', 'AvgShots', 'AvgShotsOnTarget', 'AvgCorners', 'AvgFouls', 'AvgYellowCards', 'AvgRedCards']
# Ensure all features exist in the DataFrame and handle potential infinite values
features_exist = [f for f in features if f in team_stats.columns]
X = team_stats[features_exist]
X = X.replace([np.inf, -np.inf], np.nan).dropna() # Handle infinite values and drop rows with NaNs

# Check if X is empty after dropping NaNs
if X.empty:
    print("Error: Features DataFrame is empty after handling NaNs. Cannot proceed with scaling and DBSCAN.")
else:
    # Standardisasi data
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # DBSCAN
    # Ensure min_samples is not greater than the number of samples
    min_samples_dbscan = min(2, X_scaled.shape[0])
    if min_samples_dbscan == 0:
         print("Error: Not enough samples for DBSCAN.")
    else:
        dbscan = DBSCAN(eps=1.0, min_samples=min_samples_dbscan)
        team_stats['Anomaly'] = dbscan.fit_predict(X_scaled)

        # PCA untuk visualisasi
        # Ensure n_components is not greater than the number of features or samples
        n_components_pca = min(2, X_scaled.shape[1], X_scaled.shape[0])
        if n_components_pca == 0:
             print("Error: Not enough features or samples for PCA.")
        else:
            pca = PCA(n_components=n_components_pca)
            X_pca = pca.fit_transform(X_scaled)
            team_stats['PCA1'] = X_pca[:, 0]
            if n_components_pca > 1:
                 team_stats['PCA2'] = X_pca[:, 1]
            else:
                 team_stats['PCA2'] = 0 # Add a PCA2 column even if only one component for consistent plotting

            # --- Langkah 6: Visualisasi hasil outlier ---
            plt.figure(figsize=(12, 6))
            # Map anomaly labels to colors, handling cases where no anomalies are detected (-1)
            colors = team_stats['Anomaly'].map({-1: 'red'}).fillna('blue')

            # Filter team_stats to match the rows in X_scaled used for PCA and DBSCAN
            team_stats_filtered = team_stats.loc[X.index].copy() # Use .loc to align by index
            team_stats_filtered['PCA1'] = X_pca[:, 0]
            if n_components_pca > 1:
                team_stats_filtered['PCA2'] = X_pca[:, 1]
            else:
                team_stats_filtered['PCA2'] = 0 # Use the created PCA2 column

            plt.scatter(team_stats_filtered['PCA1'], team_stats_filtered['PCA2'], c=colors, alpha=0.7)
            plt.title('Outlier Detection of Team Performance per Season (DBSCAN + PCA)')
            plt.xlabel('PCA Component 1')
            plt.ylabel('PCA Component 2' if n_components_pca > 1 else '') # Label PCA2 axis only if more than 1 component
            plt.grid(True)

            # Tandai titik anomali
            # Iterate through the filtered DataFrame
            for _, row in team_stats_filtered[team_stats_filtered['Anomaly'] == -1].iterrows():
                # Safely decode season and team names, handling potential KeyError if key not in label_encoders
                season_decoded = label_encoders.get('Season', {}).inverse_transform([row['Season']])[0] if 'Season' in label_encoders and isinstance(row['Season'], (int, np.integer)) else row['Season']
                team_decoded = label_encoders.get('HomeTeam', {}).inverse_transform([row['Team']])[0] if 'HomeTeam' in label_encoders and isinstance(row['Team'], (int, np.integer)) else row['Team']

                plt.text(row['PCA1'], row['PCA2'], f"{team_decoded} ({season_decoded})", fontsize=8, color='black')

            plt.show()
```

```python


team_stats['TeamName'] = label_encoders['HomeTeam'].inverse_transform(team_stats['Team'])
team_stats['SeasonName'] = label_encoders['Season'].inverse_transform(team_stats['Season'])

# --- Langkah 2: Filter hanya outlier ---
outliers_df = team_stats[team_stats['Anomaly'] == -1].copy()

# --- Langkah 3: Urutkan berdasarkan performa (misalnya rata-rata gol) ---
# Outlier Positif = performa sangat bagus (banyak mencetak gol)
outliers_positive = outliers_df.sort_values(by='AvgGoals', ascending=False).head(10)

# Outlier Negatif = performa sangat buruk (sedikit mencetak gol)
outliers_negative = outliers_df.sort_values(by='AvgGoals', ascending=True).head(10)

# --- Langkah 4: Tampilkan ---
print("=== 🔥 Outlier Positif (Tim dengan performa luar biasa) ===")
print(outliers_positive[['SeasonName', 'TeamName', 'AvgGoals', 'AvgShots', 'AvgShotsOnTarget']])

print("\n=== ❄️ Outlier Negatif (Tim dengan performa sangat buruk) ===")
print(outliers_negative[['SeasonName', 'TeamName', 'AvgGoals', 'AvgShots', 'AvgShotsOnTarget']])

# --- Langkah 5 (Opsional): Simpan ke file ---
outliers_df[['SeasonName', 'TeamName', 'AvgGoals', 'AvgShots', 'AvgShotsOnTarget']].to_csv("outlier_tim_perform.csv", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Salin dataframe asli
EPL_cleaned = EPL.copy()

# Label encoding untuk semua kolom bertipe objek
label_encoders = {}
for col in EPL_cleaned.select_dtypes(include='object').columns:
    EPL_cleaned[col] = EPL_cleaned[col].astype(str)
    le = LabelEncoder()
    EPL_cleaned[col] = le.fit_transform(EPL_cleaned[col])
    label_encoders[col] = le
```

```python
# Home stats
home_stats = EPL_cleaned.groupby(['Season', 'HomeTeam']).agg({
    'FullTimeHomeGoals': 'mean',
    'HomeShots': 'mean',
    'HomeShotsOnTarget': 'mean',
    'HomeCorners': 'mean',
    'HomeFouls': 'mean',
    'HomeYellowCards': 'mean',
    'HomeRedCards': 'mean'
}).reset_index().rename(columns={'HomeTeam': 'Team'})

# Away stats
away_stats = EPL_cleaned.groupby(['Season', 'AwayTeam']).agg({
    'FullTimeAwayGoals': 'mean',
    'AwayShots': 'mean',
    'AwayShotsOnTarget': 'mean',
    'AwayCorners': 'mean',
    'AwayFouls': 'mean',
    'AwayYellowCards': 'mean',
    'AwayRedCards': 'mean'
}).reset_index().rename(columns={'AwayTeam': 'Team'})

# Gabungkan
team_stats = pd.merge(home_stats, away_stats, on=['Season', 'Team'])
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Standardisasi
features = team_stats.drop(['Season', 'Team'], axis=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

# PCA (2D untuk visualisasi)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

neighbors = NearestNeighbors(n_neighbors=4)
neighbors_fit = neighbors.fit(X_scaled)
distances, indices = neighbors_fit.kneighbors(X_scaled)

# Urutkan dan plot
distances = np.sort(distances[:, 3])
plt.plot(distances)
plt.axhline(y=2.0, color='r', linestyle='--', label='Perkiraan eps')
plt.title('K-Distance Plot')
plt.xlabel('Titik data')
plt.ylabel('Jarak ke tetangga ke-4')
plt.grid(True)
plt.legend()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=2.0, min_samples=4)
cluster_labels = dbscan.fit_predict(X_scaled)

# Tambahkan ke dataframe
team_stats['Cluster'] = cluster_labels
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=team_stats['Cluster'], palette='Set1', s=100)
plt.title('Hasil Clustering DBSCAN pada Tim EPL')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```
